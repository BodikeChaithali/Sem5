```python
# ==========================================
# Volcano Countdown USING MINIMAX
# ==========================================

def minimax(state, maximizing_player):


    if state == 0:
      if maximizing_player:
        return -1
      else:
        return +1

    # 2. AI'S TURN - MAXIMIZING PLAYER


    if maximizing_player:

        # Start with the worst possible score
        best_score = -float("inf")

        # Try removing 1,2,3
        for i in range(1,4):

            if i <= state:

              new_state = state - i
              score = minimax(new_state,False)
              best_score = max(best_score,score)

        return best_score

    # 3. HUMAN'S TURN - MINIMIZING PLAYER

    else:

        # Start with the best possible score
        best_score = float("inf")

        # Try removing 1,2,3
        for i in range(1,4):

            if i <= state:
              new_state = state - i
              score = minimax(new_state,True)
              best_score = min(best_score,score)

        return best_score


def best_move(state):

    # Start with the worst possible score
    best_score = -float("inf")

    # No move selected yet
    move = None

    # Check all 3 possibilities
    for i in range(1,4):

        if i <= state:

          new_state = state - i
          score = minimax(new_state, False)

          if score > best_score:
            best_score = score
            move = i

    return move


def main():

  state = int(input("Enter Valcano CountDown: "))

  move = best_move(state)
  optimal_value = minimax(state, True)

  print("Optimal Value:", optimal_value)
  print("Best Move:", move)



# ==========================================
# START THE PROGRAM
# ==========================================

if __name__ == "__main__":
    main()
```

## Optimal Strategy for Max

Max should choose the first move that gives the highest Minimax
value. A value of +1 means Max can force a win, while -1 means
Min can prevent Max from winning.

If multiple moves have the same optimal value, the smallest move
is selected. For example, when N = 4, all three possible first
moves have value -1, so Max selects move 1.

## Minimax Complexity

At each state there can be up to 3 possible moves, and the maximum
depth of the game tree is N. Therefore, the loose worst-case time
complexity of ordinary Minimax is:

O(3^N)

## Alpha-Beta Pruning

Alpha-Beta pruning produces the same result as Minimax but avoids
evaluating branches that cannot affect the final decision.

Alpha represents the best value currently guaranteed for Max, while
beta represents the best value currently guaranteed for Min.

A branch is pruned whenever:

alpha >= beta

The assignment requires moves to be explored in the order 1, 2, 3.
The Alpha-Beta state counter counts only terminal state evaluations
where state == 0.

For the given test cases, Alpha-Beta evaluates:

N = 1 → 1 terminal state
N = 2 → 2 terminal states
N = 4 → 7 terminal states
N = 5 → 10 terminal states
N = 6 → 19 terminal states

Thus, Alpha-Beta reduces the number of terminal evaluations while
producing the same optimal result as Minimax.

```python
# ==========================================
# Volcano Countdown USING ALPHA-BETA PRUNING
# ==========================================

alpha_beta_states = 0


def alpha_beta(state, maximizing_player, alpha, beta):

    global alpha_beta_states

    # Terminal state
    if state == 0:

        alpha_beta_states += 1

        if maximizing_player:
            return -1
        else:
            return +1

    # ==========================================
    # MAXIMIZING PLAYER
    # ==========================================

    if maximizing_player:

        best_score = -float("inf")

        # Try removing 1, 2, 3
        for i in range(1, 4):

            if i <= state:

                new_state = state - i

                score = alpha_beta(
                    new_state,
                    False,
                    alpha,
                    beta
                )

                best_score = max(best_score, score)

                # Update alpha
                alpha = max(alpha, best_score)

                # Alpha-Beta pruning
                if alpha >= beta:
                    break

        return best_score

    # ==========================================
    # MINIMIZING PLAYER
    # ==========================================

    else:

        best_score = float("inf")

        # Try removing 1, 2, 3
        for i in range(1, 4):

            if i <= state:

                new_state = state - i

                score = alpha_beta(
                    new_state,
                    True,
                    alpha,
                    beta
                )

                best_score = min(best_score, score)

                # Update beta
                beta = min(beta, best_score)

                # Alpha-Beta pruning
                if alpha >= beta:
                    break

        return best_score


# ==========================================
# FIND BEST MOVE
# ==========================================

def best_move(state):

    best_score = -float("inf")
    move = None

    alpha = -float("inf")
    beta = float("inf")

    # Try moves in required order: 1, 2, 3
    for i in range(1, 4):

        if i <= state:

            new_state = state - i

            score = alpha_beta(
                new_state,
                False,
                alpha,
                beta
            )

            if score > best_score:

                best_score = score
                move = i

            # Update alpha at root
            alpha = max(alpha, best_score)

    return move, best_score


# ==========================================
# MAIN
# ==========================================

def main():

    global alpha_beta_states

    state = int(input("Enter Volcano Countdown: "))

    # Reset counter
    alpha_beta_states = 0

    # Find best move and optimal value
    move, optimal_value = best_move(state)

    print("Optimal Value:", optimal_value)
    print("Best Move:", move)
    print("Alpha-Beta States:", alpha_beta_states)


# ==========================================
# START PROGRAM
# ==========================================

if __name__ == "__main__":
    main()
```
