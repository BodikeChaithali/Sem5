```python
from collections import deque
import time
```

```python
R, C = map(int, input().split())

grid = []
for _ in range(R):
    grid.append(input().strip())

start = None
goal = None

for r in range(R):
    for c in range(C):
        if grid[r][c] == 'S':
            start = (r, c)
        elif grid[r][c] == 'G':
            goal = (r, c)

print("Start:", start)
print("Goal:", goal)
```

```python
def bfs(grid, start, goal, R, C):

    queue = deque([start])
    visited = set([start])
    parent = {start: None}
    move_taken = {}

    nodes_expanded = 0

    directions = [
        (-1, 0, "UP"),
        (0, 1, "RIGHT"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT")
    ]

    while queue:

        current = queue.popleft()
        nodes_expanded += 1

        if current == goal:
            break

        r, c = current

        for dr, dc, move in directions:

            nr = r + dr
            nc = c + dc

            if (0 <= nr < R and
                0 <= nc < C and
                grid[nr][nc] != '#' and
                (nr, nc) not in visited):

                next_state = (nr, nc)

                visited.add(next_state)
                parent[next_state] = current
                move_taken[next_state] = move

                queue.append(next_state)

    if goal not in visited:
        return None, nodes_expanded

    # Reconstruct path
    path = []
    current = goal

    while current != start:
        path.append(move_taken[current])
        current = parent[current]

    path.reverse()

    return path, nodes_expanded
```

```python
start_time = time.perf_counter()

bfs_path, bfs_nodes = bfs(grid, start, goal, R, C)

end_time = time.perf_counter()

bfs_time = end_time - start_time

print("Algorithm: BFS")

if bfs_path is None:
    print("Path Found: No")
else:
    print("Path Found: Yes")
    print("Path:", " ".join(bfs_path))
    print("Number of Moves =", len(bfs_path))
    print("Nodes Expanded =", bfs_nodes)

print("Execution Time =", bfs_time, "seconds")
```

```python
def dfs(grid, start, goal, R, C):

    stack = [start]
    visited = set([start])
    parent = {start: None}
    move_taken = {}

    nodes_expanded = 0

    directions = [
        (-1, 0, "UP"),
        (0, 1, "RIGHT"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT")
    ]

    while stack:

        current = stack.pop()
        nodes_expanded += 1

        if current == goal:
            break

        r, c = current

        # Reverse order because stack is LIFO
        for dr, dc, move in reversed(directions):

            nr = r + dr
            nc = c + dc

            if (0 <= nr < R and
                0 <= nc < C and
                grid[nr][nc] != '#' and
                (nr, nc) not in visited):

                next_state = (nr, nc)

                visited.add(next_state)
                parent[next_state] = current
                move_taken[next_state] = move

                stack.append(next_state)

    if goal not in visited:
        return None, nodes_expanded

    # Reconstruct path
    path = []
    current = goal

    while current != start:
        path.append(move_taken[current])
        current = parent[current]

    path.reverse()

    return path, nodes_expanded
```

```python
start_time = time.perf_counter()

dfs_path, dfs_nodes = dfs(grid, start, goal, R, C)

end_time = time.perf_counter()

dfs_time = end_time - start_time

print("Algorithm: DFS")

if dfs_path is None:
    print("Path Found: No")
else:
    print("Path Found: Yes")
    print("Path:", " ".join(dfs_path))
    print("Number of Moves =", len(dfs_path))
    print("Nodes Expanded =", dfs_nodes)

print("Execution Time =", dfs_time, "seconds")
```

Analysis: BFS and DFS both found a path of 8 moves for Sample 1. BFS expanded 19 nodes, while DFS expanded 11 nodes. Although DFS expanded fewer nodes in this particular experiment, its execution time was slightly higher. This difference can occur because the grid is very small and execution times are affected by system and Python overhead. BFS guarantees an optimal shortest path because all movements have equal cost, whereas DFS does not guarantee an optimal path. Therefore, BFS is preferable when the shortest path is required.
