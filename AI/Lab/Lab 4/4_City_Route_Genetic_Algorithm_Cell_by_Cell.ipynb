# 4-City Delivery Route Optimization using Genetic Algorithm

A delivery person starts at **A**, visits **B, C, and D** exactly once, and returns to **A**. The objective is to minimize total travel cost.

```python
# Take the problem input

def take_input():
    n = int(input("Enter number of locations: "))

    locations = input("Enter locations separated by spaces: ").upper().split()

    start = input("Enter starting location: ").upper()

    costs = []

    print("Enter cost matrix:")

    for i in range(n):
        row = list(map(int, input().split()))
        costs.append(row)

    return n, locations, start, costs

n, locations, start, costs = take_input()

print("Number of locations:", n)
print("Locations:", locations)
print("Starting location:", start)
```

## Step 2: Create the initial population

Each chromosome represents a valid route. The starting location is fixed and the remaining locations are shuffled.

```python
# Create initial population

import random

def create_population(size):
    population = []

    remaining = [city for city in locations if city != start]

    for _ in range(size):
        chromosome = remaining.copy()
        random.shuffle(chromosome)
        population.append(chromosome)

    return population

population_size = 6
population = create_population(population_size)

print("Initial Population:")
for route in population:
    print(" → ".join([start] + route + [start]))
```

## Step 3: Calculate the route cost

The fitness of a route is its total travel cost. A lower cost represents a better route.

```python
# Calculate route cost

def route_cost(route):
    full_route = [start] + route + [start]
    total = 0

    for i in range(len(full_route) - 1):
        from_city = locations.index(full_route[i])
        to_city = locations.index(full_route[i + 1])
        total += costs[from_city][to_city]

    return total

print("Population costs:")
for route in population:
    print(" → ".join([start] + route + [start]),
          "| Cost:", route_cost(route))
```

## Step 4: Select parent routes

Two routes are selected randomly and the route with the lower travel cost is selected as the parent.

```python
# Select parent

def select_parent(population):
    candidates = random.sample(population, 2)

    if route_cost(candidates[0]) < route_cost(candidates[1]):
        return candidates[0].copy()

    return candidates[1].copy()
```

## Step 5: Apply crossover

Crossover combines two parent routes while keeping every location exactly once.

```python
# Crossover

def crossover(parent1, parent2):
    size = len(parent1)

    point1, point2 = sorted(random.sample(range(size), 2))

    child = [None] * size
    child[point1:point2 + 1] = parent1[point1:point2 + 1]

    remaining = [city for city in parent2 if city not in child]

    j = 0

    for i in range(size):
        if child[i] is None:
            child[i] = remaining[j]
            j += 1

    return child
```

## Step 6: Apply mutation

Mutation swaps two locations in a route.

```python
# Mutation

def mutate(route):
    if random.random() < 0.2:
        i, j = random.sample(range(len(route)), 2)
        route[i], route[j] = route[j], route[i]

    return route
```

## Step 7: Run the Genetic Algorithm

```python
# Genetic Algorithm

def genetic_algorithm(population):
    best_route = min(population, key=route_cost).copy()

    for generation in range(20):
        new_population = []

        while len(new_population) < population_size:
            parent1 = select_parent(population)
            parent2 = select_parent(population)

            child = crossover(parent1, parent2)
            child = mutate(child)

            new_population.append(child)

        population = new_population

        current_best = min(population, key=route_cost)

        if route_cost(current_best) < route_cost(best_route):
            best_route = current_best.copy()

    return best_route

solution = genetic_algorithm(population)

print("Best Route:", " → ".join([start] + solution + [start]))
print("Minimum Total Cost:", route_cost(solution))
```

```python
# Display the final solution

print("Final solution:")
print("A → " + " → ".join(solution) + " → A")
print("Minimum cost:", route_cost(solution))
```
