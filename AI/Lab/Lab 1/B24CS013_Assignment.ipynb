```python
def depth_limited_search(graph,current,goal,limit,path,visited_order):
  visited_order.append(current)
  path.append(current)
  if current == goal:
    return True

  if limit == 0:
    path.pop()
    return False

  for neighbor in graph[current]:
    if neighbor not in path:
      if depth_limited_search(graph,neighbor,goal,limit - 1,path,visited_order):
        return True

  path.pop()
  return False

def iterative_deepening_search(graph,start,goal):
  max_depth = len(graph)
  for depth in range(max_depth+1):
    visited_order = []
    path = []

    print(f"\nDepth Limit = {depth}")

    found = depth_limited_search(graph, start, goal, depth, path, visited_order)

    print("Nodes Visited:", " -> ".join(visited_order))

    if found:
      print("Goal Found!")
      print("Path:", " -> ".join(path))
      return

    print("Goal not found at this depth.")

# -----input-----

graph = {}

n = int(input("Enter number of nodes: "))

print("\nEnter the node names:")
for i in range(n):
    node = input(f"Node {i+1}: ")
    graph[node] = []

e = int(input("\nEnter number of edges: "))

print("\nEnter each edge as: source destination")
for i in range(e):
    u, v = input(f"Edge {i+1}: ").split()

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)


# -------------------------------
# Display Graphh--------------
# Starting Node
# -------------------------------

start = input("\nEnter starting node for IDS: ")
goal = input("\nEnter goal node for IDS: ")

print("\nIterative Deepening Search")
iterative_deepening_search(graph, start, goal)
```
